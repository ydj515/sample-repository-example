```kotlin
import com.example.kotlinnotebookexample.KotlinNotebookExampleApplication
import org.springframework.boot.builder.SpringApplicationBuilder

// (선택) 스프링부트 로깅 끄기
System.setProperty("org.springframework.boot.logging.LoggingSystem", "none")

val ctx = SpringApplicationBuilder(KotlinNotebookExampleApplication::class.java)
    .properties(
        mapOf(
            "server.port" to 0 // 빈 포트 자동 배정
        )
    )
    .run()

val port = ctx.environment.getProperty("local.server.port")
println("started on http://localhost:$port")
```

```kotlin
import java.net.URI
import java.net.http.HttpClient
import java.net.http.HttpRequest
import java.net.http.HttpResponse

val client = HttpClient.newHttpClient()

val request = HttpRequest.newBuilder()
    .uri(URI.create("http://localhost:$port/hello"))
    .GET()
    .build()

val response = client.send(request, HttpResponse.BodyHandlers.ofString())

println("status = ${response.statusCode()}")
println("body   = ${response.body()}")
```

```kotlin
ctx.close()
```
